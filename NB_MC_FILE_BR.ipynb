# Load file to bronze

## Imports

```python
import os
from datetime import datetime
import pandas as pd
```

## Parameters

```python
source_file_path = os.path.join(os.curdir, "menstrual_cycle_editado.csv")
```

```python
os.listdir()
```

```python
bronze_file_name = "menstrual_cycle_bronze.parquet"
bronze_path = os.path.join(os.curdir, "Data_Lake","Bronze")
bronze_file_path = os.path.join(bronze_path, bronze_file_name)
```

```python
os.makedirs(bronze_path, exist_ok = True)
```

## Main

### Cargar el csv a un DataFrame

```python
menstrual_cycle_df = pd.read_csv(source_file_path,sep=";")
```

```python
# Comprobamos si el fichero se ha cargado correctamente
menstrual_cycle_df.head()
```

### Añadir el timestamp de carga

```python
menstrual_cycle_df['_BronzeTimestamp'] = datetime.now()
```

```python
menstrual_cycle_df.head()
```

### Guardar el DataFrame en parquet

```python
menstrual_cycle_df.to_parquet(bronze_file_path, index=False)
```
